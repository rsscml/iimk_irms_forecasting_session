## Notebook 3 · LightGBM for Forecasting — One-Step (Recursive) Model

**Goals**

1. Train a single **global** LightGBM model that predicts $y_{t+1}$ for every forecast key.
2. Choose the **objective** (Tweedie / Poisson / RMSE) given the demand mix from Notebook 1.
3. Use **early stopping** with a time-based validation set.
4. Generate forecasts for an entire test horizon by **recursive roll-out**.
5. Evaluate per key and in aggregate.

The next notebook contrasts this with the Direct Multi-Horizon strategy and a Purely-Causal variant.

```python
# === Dataset configuration ==================================================
# EDIT THIS CELL TO POINT AT YOUR DATASET
DATA_PATH    = "./dataset/m5/m5_small.csv"             # path to your CSV / parquet
DATE_COL     = "date"                              # date column
TARGET_COL   = "sales"                          # target / forecast column
KEY_COLS     = ['item_id',
                'dept_id',
                'cat_id',
                'store_id',
                'state_id']                           # columns identifying a unique series
FREQ         = "D"                              # 'D'=daily, 'W'=weekly, 'M'=monthly
HOLDOUT_DAYS = 28                               # length of test horizon
```

```python
from utils.data_utils import load_forecasting_data, time_based_split, classify_demand
from utils.feature_engineering import FeatureEngineer

df = load_forecasting_data(DATA_PATH, DATE_COL, TARGET_COL, KEY_COLS)
cutoff = df[DATE_COL].max() - pd.Timedelta(days=HOLDOUT_DAYS - 1)
train_df, test_df = time_based_split(df, DATE_COL, cutoff)

# We carve out the LAST `HOLDOUT_DAYS` of TRAIN as a tuning validation set,
# leaving the original test_df untouched as the FINAL holdout.
val_cutoff = train_df[DATE_COL].max() - pd.Timedelta(days=HOLDOUT_DAYS - 1)
fit_df, val_df = time_based_split(train_df, DATE_COL, val_cutoff)
print(f"  fit  : {fit_df[DATE_COL].min().date()} → {fit_df[DATE_COL].max().date()}  ({len(fit_df):,} rows)")
print(f"  val  : {val_df[DATE_COL].min().date()} → {val_df[DATE_COL].max().date()}  ({len(val_df):,} rows)")
print(f"  test : {test_df[DATE_COL].min().date()} → {test_df[DATE_COL].max().date()}  ({len(test_df):,} rows)")
```

### 1. Choose the objective from the demand mix


```python
from utils.losses import recommend_loss

cls = classify_demand(df, DATE_COL, KEY_COLS, TARGET_COL)
mode_pattern = cls["pattern"].value_counts().idxmax()
objective = recommend_loss(mode_pattern)
print(f"Dominant pattern: {mode_pattern}  ->  LightGBM objective: '{objective}'")
```

### 2. Configure the feature engineer and the forecaster

The `RecursiveForecaster` trains **one** model that predicts a single step ahead. At inference, we roll the prediction forward day-by-day, feeding each new prediction back into the lag features.


```python
from utils.feature_engineering import FeatureEngineer
from utils.lgbm_forecaster import RecursiveForecaster

fe = FeatureEngineer(
    date_col=DATE_COL, target_col=TARGET_COL, key_cols=KEY_COLS,
    lags=[1, 2, 7, 14, 28],
    rolling_windows=[7, 14, 28],
    rolling_stats=["mean", "std", "max"],
    ewm_halflives=[7.0, 28.0],
)

params = {
    "objective": objective,                 # 'tweedie' / 'poisson' / 'regression'
    "metric": "rmse",
    "learning_rate": 0.05,
    "num_leaves": 63,
    "min_data_in_leaf": 50,
    "feature_fraction": 0.8,
    "bagging_fraction": 0.8,
    "bagging_freq": 1,
    "lambda_l2": 0.1,
    "verbose": -1,
    "verbose_eval": 10,
}
if objective == "tweedie":
    params["tweedie_variance_power"] = 1.2

forecaster = RecursiveForecaster(
    feature_engineer=fe,
    params=params,
    num_boost_round=2000,
    early_stopping_rounds=10,
)
```

### 3. Train

We pass `valid_df=val_df` so LightGBM can stop early when the validation metric plateaus.


```python
forecaster.fit(fit_df, valid_df=val_df)
print(f"best iteration : {forecaster.model.best_iteration}")
print(f"# features     : {len(forecaster.feature_names)}")
```

### 4. Inspect feature importance (gain)

A first look at *which* features the model relies on. We will return to this in Notebook 7 with proper SHAP attributions.


```python
from utils.explainer import LGBMExplainer
from utils.viz import plot_importance

ex = LGBMExplainer(forecaster.model, feature_names=forecaster.feature_names)
imp = ex.importance(kind="gain")
plot_importance(imp, top_n=20, value_col="importance",
                title="Top-20 features by gain (recursive model)").show()
```

### 5. Predict the holdout horizon — recursive roll-out

For each key we need the most recent `max(lag, window)` days as **history**. The forecaster appends a row for each future date, recomputes lag/rolling features, predicts, and feeds the prediction into the next step.


```python
train_df.columns.tolist()
```

```python
horizon_dates = pd.date_range(test_df[DATE_COL].min(), test_df[DATE_COL].max(), freq=FREQ)


# Dataset Specific features
holiday_cols = [c for c in train_df.columns if c.startswith("event_")]
snap_cols = [c for c in train_df.columns if c.startswith("snap_")]
other_future_known_cols = ['sell_price']

exo_future = test_df[[DATE_COL] + KEY_COLS + holiday_cols + snap_cols + other_future_known_cols].copy()


# History needs to include enough lag context. Easiest: pass the entire training set.
preds = forecaster.predict_recursive(
    history_df=train_df,                # everything BEFORE the test window
    horizon_dates=horizon_dates,
    exogenous_future=exo_future,
)
preds.head()
```

### 6. Evaluate


```python
from utils.metrics import metric_report, metric_report_by_key

# Join predictions with actuals
joined = (test_df[[DATE_COL, *KEY_COLS, TARGET_COL]]
          .merge(preds, on=[DATE_COL, *KEY_COLS], how="inner"))

print("Aggregate metrics on the test horizon:")
print(metric_report(joined[TARGET_COL].values, joined["yhat"].values).round(3))

per_key = metric_report_by_key(joined, KEY_COLS, TARGET_COL, "yhat")
per_key.head()
```

### 7. Visualise — best, worst, median key


```python
from utils.viz import plot_forecast

per_key_sorted = per_key.sort_values("WAPE%")
best_key   = per_key_sorted.iloc[0][KEY_COLS].tolist()
median_key = per_key_sorted.iloc[len(per_key_sorted)//2][KEY_COLS].tolist()
worst_key  = per_key_sorted.iloc[-1][KEY_COLS].tolist()

def select(df_in, key_values):
    out = df_in.copy()
    for c, v in zip(KEY_COLS, key_values):
        out = out[out[c] == v]
    return out

for label, key in [("best", best_key), ("median", median_key), ("worst", worst_key)]:
    hist = select(train_df, key).tail(120)
    act  = select(test_df, key)
    pre  = select(preds, key)
    fig = plot_forecast(act, pre, DATE_COL, TARGET_COL, history=hist,
                        title=f"{label.upper()} key — {dict(zip(KEY_COLS, key))}")
    fig.show()
```

### 8. Residual diagnostics

Are errors centred around zero? Heteroscedastic? Plotly residual triplet:


```python
from utils.viz import plot_residuals
plot_residuals(joined[TARGET_COL].values, joined["yhat"].values).show()
```
