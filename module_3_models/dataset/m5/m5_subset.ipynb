```python
df['unique_id'] = df['item_id'] + '_' + df['store_id']

print(df['unique_id'].nunique())

sampled_keys = pd.Series(df["unique_id"].unique()).sample(n=500, random_state=42)

df_subset = df[df["unique_id"].isin(sampled_keys)].copy()
```

```python
print(df_subset['unique_id'].nunique())
```

```python
df_subset.drop(columns=['unique_id'], inplace=True)
```

```python
df_subset['date'] = pd.to_datetime(df_subset['date'])
```

```python
df_subset = df_subset[df_subset['date']<=pd.to_datetime("2016-05-20",format="%Y-%m-%d")]
```

```python
df_subset = df_subset.reset_index(drop=True)
```
