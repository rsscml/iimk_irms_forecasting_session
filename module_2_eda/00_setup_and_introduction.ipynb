## What is forecasting?
Forecasting is the task of predicting future values of a quantity that is
indexed by **time** — daily store sales, weekly call-centre volume, hourly
energy consumption. A forecasting dataset typically has three pieces:

- **Time** — a regularly spaced (or supposed-to-be-regularly-spaced) index.
- **Target** — what we want to predict (sales, demand, revenue).
- **Forecast keys** — columns that, in combination, identify *one* series
  (for example, `store_id × item_id` gives one series per SKU per store).

Most real datasets contain **many** series at once (thousands or millions),
each with its own length, its own pattern, and its own quirks. This is why
robust *data preparation* matters more than picking a fancy model — and why
this whole tutorial is dedicated to it.

## The DatasetSpec idea

Every notebook in this series uses the same lightweight pattern: a
**DatasetSpec** dictionary that declares the role of each column. The
toolkit's functions read this spec, so the same code works for the
*Walmart M5* dataset, the *Rossmann* dataset, *Store Item Demand*, or any
private dataset you bring.

```python
spec = ft.data_io.make_spec(
    date_col='date',
    target_col='sales',
    key_cols=['store_id', 'item_id'],
    static_cols=['store_type', 'category'],
    dynamic_cols=['promo', 'price'],
    frequency='D',
)
```

Pick **one** dataset to use for the rest of the session. Common choices:
